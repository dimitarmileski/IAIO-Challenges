```python
import pandas as pd
import numpy as np
from sklearn.metrics import mean_squared_error
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
```

```python
train_data = pd.read_csv('train.csv')
```

```python
train_data
```

```python
train_data.info()
```

```python
columns_missing_values = [column for column in train_data.columns if train_data[column].notnull().sum() < 3000]
columns_missing_values
```

```python
train_data_after_dropping = train_data.copy()
train_data_after_dropping = train_data.drop(columns = columns_missing_values)
```

```python
train_data_after_dropping
```

```python
train_data_after_dropping.info()
```

```python
from sklearn.impute import SimpleImputer
```

```python
train_data_after_dropping
```

```python
for i in train_data_after_dropping.columns:
    if train_data_after_dropping[i].dtype == 'object':
        print(i, train_data_after_dropping[i].isnull().sum())
```

```python
train_data_after_dropping['RcNursEstDate'] = pd.to_datetime(train_data_after_dropping['RcNursEstDate'])
```

```python
columns_missing_greater_than_0 = []
for i in train_data_after_dropping.columns:
    if train_data_after_dropping[i].dtype == 'object':
        if train_data_after_dropping[i].isnull().sum() > 0:
            columns_missing_greater_than_0.append(i)
columns_missing_greater_than_0
```

```python
categorical_imputer = SimpleImputer(strategy='most_frequent')
train_data_after_imputation = train_data_after_dropping.copy()

train_data_after_imputation[columns_missing_greater_than_0] = categorical_imputer.fit_transform(
    train_data_after_dropping[columns_missing_greater_than_0]
)
```

```python
train_data_after_imputation.info()
```

```python
train_data_after_imputation = train_data_after_imputation.drop(columns = ['RcNursEstDate'])
```

```python
for i in train_data_after_imputation.columns:
    if train_data_after_imputation[i].dtype == 'object':
        print(i, train_data_after_imputation[i].nunique())
```

```python
train_data_after_imputation['CropTillageDate'] = pd.to_datetime(train_data_after_imputation['CropTillageDate'])
train_data_after_imputation['Harv_date'] = pd.to_datetime(train_data_after_imputation['Harv_date'])
train_data_after_imputation['Threshing_date'] = pd.to_datetime(train_data_after_imputation['Threshing_date'])
```

```python
train_data_after_imputation.info()
```

```python
encoder = LabelEncoder()
columns_for_encoding = []
for i in train_data_after_imputation.columns:
    if train_data_after_imputation[i].dtype == 'object' and i != 'ID':
        columns_for_encoding.append(i)
train_data_after_encoding = train_data_after_imputation.copy()
for i in columns_for_encoding:
    train_data_after_encoding[i] = encoder.fit_transform(train_data_after_imputation[i])
```

```python
train_data_after_encoding
```

```python
train_data_after_encoding.info()
```

```python
cols_for_imputation = []
for i in train_data_after_encoding.columns:
    if train_data_after_encoding[i].isnull().sum() > 0:
        cols_for_imputation.append(i)
cols_for_imputation
```

```python
imputer = SimpleImputer(strategy = 'median')
dataset = train_data_after_encoding.copy()
dataset[cols_for_imputation] = imputer.fit_transform(train_data_after_encoding[cols_for_imputation])
```

```python
dataset
```

```python
dataset.info()
```

```python
dataset = dataset.drop(columns = ['CropTillageDate', 'Harv_date', 'Threshing_date'])
```

```python
label = dataset['Yield']
features = dataset.drop(columns = ['ID', 'Yield', 'Block', 'Stubble_use', '1tdUrea'])
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.svm import SVR
from sklearn.tree import DecisionTreeRegressor
from sklearn.neural_network import MLPRegressor
from sklearn.ensemble import GradientBoostingRegressor, RandomForestRegressor
```

```python
(features_train, features_test, label_train, label_test) = train_test_split(features, label, test_size = 0.2)
```

```python
import numpy as np
lin_reg = LinearRegression()
lin_reg.fit(features_train, label_train)
lin_reg_score = lin_reg.score(features_test, label_test)
lin_reg_predicted = lin_reg.predict(features_test)
lin_reg_rmse = np.sqrt(mean_squared_error(label_test, lin_reg_predicted))
print('Linear regression accuracy:', lin_reg_score, 'rmse:', lin_reg_rmse)
```

```python
svr = SVR()
svr.fit(features_train, label_train)
svr_score = svr.score(features_test, label_test)
svr_predicted = svr.predict(features_test)
svr_rmse = np.sqrt(mean_squared_error(label_test, svr_predicted))
print('SVR accuracy:', svr_score, 'rmse:', svr_rmse)
```

```python
tree = DecisionTreeRegressor()
tree.fit(features_train, label_train)
tree_score = tree.score(features_test, label_test)
tree_predicted = tree.predict(features_test)
tree_rmse = np.sqrt(mean_squared_error(label_test, tree_predicted))
print('tree accuracy:', svr_score, 'rmse:', svr_rmse)
```
